# Errors and Exceptions in Python

There are two main error categories in Python: **syntax errors** (code that breaks Python grammar) and **exceptions** (runtime errors that occur while executing otherwise valid code). Handling exceptions lets programs fail gracefully instead of crashing.

## Syntax Errors

These are parsing errors where Python cannot interpret the code. Example:

```python
this is not valid code, so it will error
```
Python will point to the invalid syntax and stop execution.

## Exceptions

An exception is a signal that something unexpected happened while your program was running. Python stops normal execution and looks for code that can handle the problem. You can handle these with `try`/`except` to keep programs from crashing and to respond gracefully.

```python
try:
    10 / 0
except Exception:
    print("can't divide by zero")
```

To access the exception details, bind it with `as`:

```python
try:
    10 / 0
except Exception as e:
    print(e)  # prints: division by zero
```

Wrap risky operations in `try`/`except` when you want controlled recovery instead of an abrupt crash.

````python
def main():

    try:
        print(get_player_record(1))
        print(get_player_record(2))
        print(get_player_record(3))
        print(get_player_record(4))
        print(get_player_record(5))
    except Exception as e:
        print(e)
        



# Don't edit below this line


def get_player_record(player_id):
    if player_id == 1:
        return {"name": "Slayer", "level": 128}
    if player_id == 2:
        return {"name": "Dorgoth", "level": 300}
    if player_id == 3:
        return {"name": "Saruman", "level": 4000}
    if player_id == 5:
        return {"name": "Gandalf", "level": 5000}
    raise Exception("player id not found")


main()
````

## Different Types of Exceptions
We haven't covered classes and objects yet, which is what an Exception really is at its core. We'll go more into that in the course on object-oriented programming.

For now, what is important to understand is that there are different types of exceptions, and we can handle them differently depending on the situation. Some exceptions are more specific, like ZeroDivisionError (which happens when you divide by zero) or IndexError (which happens when you try to access a list element at an invalid index—either too high or too low). Others are more general, like the base Exception.

### Syntax
```python
try:
    10/0
except ZeroDivisionError:
    print("0 division")
except Exception as e:
    print(e)

try:
    nums = [0, 1]
    print(nums[2])
except IndexError:
    print("index error")
except Exception as e:
    print(e)
```

Which will print:

0 division
index error

### Why Specific Exceptions Come First
When handling exceptions, it’s important to catch the most specific ones first, because Python stops checking once it finds a matching exception handler. If you catch a more general Exception first, any specific errors will never get handled individually.

For example:

```python
try:
    nums = [0, 1]
    print(nums[2])
except Exception:
    print("An error occurred")
except IndexError:
    print("Index error")
```

In this case, the general Exception will catch the error before the IndexError can be reached, and the message “Index error” will never be printed. Always handle the most specific exception first!

### Alias Exception Messages
As seen in the example, you can also access the error using as, like this:

```python
except Exception as e:
    print(e)
```

The default behavior of print is that it will print the string representation of whatever object is passed to it. In this case, it will print the error message.
