# Sets in Python

Sets are unordered collections that automatically keep only unique elements. Use them when membership checks and deduplication matter more than order.

# Basic set creation and uniqueness

Sets remove duplicates automatically and do not preserve order.

```python
# Create a set and inspect its type
fruits = {"apple", "banana", "grape"}
print(type(fruits))
print(fruits)  # order may vary
```

# Add values to a set

Use `.add()` to insert items; duplicates are ignored.

```python
fruits = {"apple", "banana", "grape"}
fruits.add("pear")
fruits.add("apple")  # duplicate ignored
print(fruits)
```

# Creating an empty set

Use `set()` because `{}` makes a dictionary.

```python
fruits = set()
fruits.add("pear")
print(fruits)
```

# Iterating over a set

Order is not guaranteed during iteration.

```python
fruits = {"apple", "banana", "grape"}
for fruit in fruits:
    print(fruit)
```

# Removing values (again)

`.remove()` raises `KeyError` if the item is missing; `.discard()` quietly does nothing when the item is absent.

```python
fruits = {"apple", "banana", "grape"}
fruits.remove("apple")
print(fruits)  # {'banana', 'grape'}

# Using discard to avoid KeyError
fruits.discard("pear")  # no error even if missing
print(fruits)
```

# Set subtraction (difference)

Using `-` removes all elements of the right-hand set from the left-hand set.

```python
set1 = {"apple", "banana", "grape"}
set2 = {"apple", "banana"}
set3 = set1 - set2
print(set3)  # {'grape'}
```
