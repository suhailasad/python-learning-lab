# Dictionaries
Dictionaries store data as `key: value` pairs. Use curly braces `{}` and unique, hashable keys. Values can be any type. Access with `dict[key]` or `dict.get(key)` to avoid `KeyError`.

**Common methods**: `.keys()`, `.values()`, `.items()`, `.get(key, default)`, `.pop(key)`, `in` for membership (checks keys).


````python
# create a dictionary
car = {
    "brand": "Toyota",
    "model": "Camry",
    "year": 2019,
}

car
````

````python
# access, add/update, remove, iterate
print("brand:", car["brand"])
print("year via get:", car.get("year"))

# add/update
car["color"] = "Blue"
car["year"] = 2020

# remove
removed_model = car.pop("model") # or #  del car["model"]


# iterate
for key, value in car.items():
    print(f"{key}: {value}")

print("removed model:", removed_model)
````

````python
def get_character_record(name, server, level, rank):

    character_record = {
        "name": name,
        "server": server,
        "level": level,
        "rank": rank,
        "id" : f"{name}#{server}",
    }

    return character_record
````

## Duplicate keys overwrite

Because dictionary keys must be unique, assigning the same key again overwrites the previous value rather than creating a second entry.


```python
example = {"color": "blue", "color": "red"}
print(example)  # {'color': 'red'}
```

Use distinct keys or namespaces (e.g., nested dicts) when you need separate values.

## Checking for existence
Use `in` to test whether a key exists before accessing it.

```python
cars = {
    "ford": "f150",
    "toyota": "camry",
}

print("ford" in cars)   # True
print("gmc" in cars)    # False
```


## Deleting dictionary values
Use `del` to remove an existing key.

```python
names_dict = {
    "jack": "bronson",
    "jill": "mcarty",
    "joe": "denver",
}

# delete an existing key
del names_dict["joe"]

print(names_dict)
# {'jack': 'bronson', 'jill': 'mcarty'}
```

Attempting to delete a missing key raises a `KeyError`.

```python
names_dict = {
    "jack": "bronson",
    "jill": "mcarty",
    "joe": "denver",
}

del names_dict["unknown"]  # KeyError: 'unknown'
```

Use `if key in dict` or `.pop(key, default)` to avoid the error when unsure a key exists.

## Iterating over a dictionary
Iterate directly over keys and then access their values.

```python
fruit_sizes = {
    "apple": "small",
    "banana": "large",
    "grape": "tiny",
}

for name in fruit_sizes:
    size = fruit_sizes[name]
    print(f"name: {name}, size: {size}")
```


## Iteration helpers (.items, .values, .keys, enumerate)
Handy ways to loop through dictionaries depending on what you need.

```python
fruit_sizes = {
    "apple": "small",
    "banana": "large",
    "grape": "tiny",
}

# 1) Keys and values together
for name, size in fruit_sizes.items():
    print(f"name: {name}, size: {size}")

# 2) Just values
for size in fruit_sizes.values():
    print(f"size: {size}")

# 3) Just keys
for name in fruit_sizes.keys():
    print(f"name: {name}")

# 4) With index + pairs
for i, (name, size) in enumerate(fruit_sizes.items()):
    print(f"{i}: name: {name}, size: {size}")
```


## Merging dictionaries (later keys win)
When keys overlap, the later dictionary’s values overwrite earlier ones.

```python
two_towers = {"Frodo": 56, "Aragorn": 10}
rotk = {"Aragorn": 100, "Gandalf": 809}

def merge(first, second):
    merged = first.copy()
    merged.update(second)  # second wins on conflicts
    return merged

merged_dict = merge(two_towers, rotk)
print(merged_dict)
# {'Frodo': 56, 'Aragorn': 100, 'Gandalf': 809}
```
