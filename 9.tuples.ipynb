# Tuples
Tuples are ordered, immutable collections with fixed size, created with parentheses:

```python
my_tuple = ("this is a tuple", 45, True)
print(my_tuple[0])  # this is a tuple
print(my_tuple[1])  # 45
print(my_tuple[2])  # True
```

While mixing types in a list is usually avoided, it is common in tuples because their positions are fixed and predictable. Tuples often hold just a few related values, like a name and age:

```python
dog = ("Fido", 4)
```

Single-item tuples need a trailing comma so Python knows it is a tuple:

```python
dog = ("Fido",)
```

You can store tuples inside a list. The first index selects the tuple; the second selects a value inside that tuple:

```python
my_tuples = [
    ("this is the first tuple in the list", 45, True),
    ("this is the second tuple in the list", 21, False)
]
print(my_tuples[0][0])  # this is the first tuple in the list
print(my_tuples[0][1])  # 45
print(my_tuples[1][0])  # this is the second tuple in the list
print(my_tuples[1][2])  # False
```

## Tuple Unpacking
Unpacking assigns tuple values to variables in one step:

```python
dog = ("Fido", 4)
dog_name, dog_age = dog
print(dog_name)  # Fido
print(dog_age)   # 4
```

Returning multiple values from a function actually returns a tuple.
