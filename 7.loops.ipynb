### Using the optional `step` in `range()`



- `range(start, stop, step)` lets you control the increment (or decrement) each loop.

- Positive step moves forward; negative step lets you count backwards.



```python
for i in range(0, 10, 2):
    print(i)
# 0
# 2
# 4
# 6
# 8
```



```python
for i in range(3, 0, -1):
    print(i)
# 3
# 2
# 1
```

````python
for i in range(3, 5, -2):
    print(i)

"""
Why it runs with no error

range(start, stop, step) is valid as long as step ≠ 0; the sign of step just determines direction.
With range(3, 5, -2), you ask to move downward from 3 but the stop (5) is above the start, so the sequence is empty.
Empty range → loop body never executes → no prints, no error.
"""
````

````python
def count_down(start, end):
    for i in range(start, end, -1):
        print(i)


# Don't edit below this line


def test(start, end):
    print(f"Using inputs start: {start} and end: {end}")
    print(f"Printing numbers from {start} to {end + 1}:")
    count_down(start, end)
    print("=====================================")


def main():
    test(10, 0)
    test(20, 10)
    test(15, 11)


main()
````

````python
def countdown_to_start():
    for i in range(10, 0, -1):
        if i == 1:
            print("1...Fight!")
        else:
            print(f"{i}...")


# Don't edit below this line


def test():
    print("Counting down to match start:")
    countdown_to_start()
    print("=====================================")


def main():
    test()


main()
````

### Using in-place operators (`+=`, `-=`)

- `variable += amount` increases the current value by `amount`.

- `variable -= amount` decreases the current value by `amount`.



```python
number_of_enemies = 10
number_of_enemies += 2  # now 12

number_of_enemies = 10
number_of_enemies -= 2  # now 8
```


````python
# main.py
def sum_of_numbers(start, end):
    total = 0
    for i in range(start, end):
        total += i
    return total
````

````python
# main_test.py
from main import *

run_cases = [(0, 5, 10), (0, 10, 45), (10, 20, 145)]

submit_cases = run_cases + [
    (1, 100, 4950),
    (5, 50, 1215),
    (20, 30, 245),
    (50, 60, 545),
    (100, 110, 1045),
]


def test(input1, input2, expected_output):
    print("---------------------------------")
    print(f"Inputs:")
    print(f" * start: {input1}")
    print(f" * end: {input2}")
    result = sum_of_numbers(input1, input2)
    print(f"Expected: {expected_output}")
    print(f"Actual:   {result}")
    if result == expected_output:
        print("Pass")
        return True
    print("Fail")
    return False


def main():
    passed = 0
    failed = 0
    skipped = len(submit_cases) - len(test_cases)
    for test_case in test_cases:
        correct = test(*test_case)
        if correct:
            passed += 1
        else:
            failed += 1
    if failed == 0:
        print("============= PASS ==============")
    else:
        print("============= FAIL ==============")
    if skipped > 0:
        print(f"{passed} passed, {failed} failed, {skipped} skipped")
    else:
        print(f"{passed} passed, {failed} failed")


test_cases = submit_cases
if "__RUN__" in globals():
    test_cases = run_cases

main()
````

### Using `while` loops
- Runs repeatedly *while* a condition is `True`.
- Commonly used when the number of iterations isn’t known in advance.

```python
while 1:
    print("1 evaluates to True")
# infinite loop (condition always True)
```

```python
num = 0
while num < 3:
    num += 1
    print(num)
# 1
# 2
# 3  (stops when num >= 3)
```


````python
# main.py
def regenerate(current_health, max_health, enemy_distance):
    while current_health < max_health and enemy_distance > 3:
        current_health += 1
        enemy_distance -= 2
    return current_health
````

````python
#main_test.py
from main import *

run_cases = [(0, 10, 20, 9), (0, 10, 4, 1), (8, 10, 20, 10)]

submit_cases = run_cases + [
    (0, 0, 0, 0),
    (9, 10, 3, 9),
    (100, 100, 200, 100),
    (2, 110, 50, 26),
    (100, 1010, 2000, 1010),
]


def test(input1, input2, input3, expected_output):
    print("---------------------------------")
    print("Inputs:")
    print(f" * current_health: {input1}")
    print(f" *     max_health: {input2}")
    print(f" * enemy_distance: {input3}")
    print(f"Expected Health: {expected_output}")
    result = regenerate(input1, input2, input3)
    print(f"  Actual Health: {result}")
    if result == expected_output:
        print("Pass")
        return True
    print("Fail")
    return False


def main():
    passed = 0
    failed = 0
    skipped = len(submit_cases) - len(test_cases)
    for test_case in test_cases:
        correct = test(*test_case)
        if correct:
            passed += 1
        else:
            failed += 1
    if failed == 0:
        print("============= PASS ==============")
    else:
        print("============= FAIL ==============")
    if skipped > 0:
        print(f"{passed} passed, {failed} failed, {skipped} skipped")
    else:
        print(f"{passed} passed, {failed} failed")


test_cases = submit_cases
if "__RUN__" in globals():
    test_cases = run_cases

main()
````

### Using `continue` to skip iterations
- `continue` jumps to the next loop iteration, skipping the rest of the body for the current one.
- Example: print 1–50 but skip every 7th number.

```python
counter = 0
for number in range(1, 51):  # 1 through 50
    counter += 1

    if counter == 7:
        counter = 0  # reset the counter
        continue     # skip printing this number

    print(number)
```


### `continue` to avoid unnecessary work
- Skip iterations that you don’t need to process.
- Example: skip negatives when taking square roots.

```python
numbers = [16, -4, 25, -9, 36, 0, 49]

for number in numbers:
    if number < 0:
        continue  # skip negatives to avoid complex results
    print(f"The square root of {number} is {number ** 0.5}.")
```

Using `continue` this way can speed up loops by bypassing needless work.


### Using `pass` (no-op)
- Placeholder when syntax requires a statement but you want to do nothing.
- Useful for stubs, ignored branches, or intentionally ignored exceptions.

```python
# Stub to fill later
def todo_feature():
    pass
```

```python
# Branch you intentionally ignore
for x in items:
    if x < 0:
        pass  # explicitly do nothing
    else:
        handle(x)
```

```python
# Ignore specific exception cases
try:
    risky()
except SomeError:
    pass  # intentionally ignore
```


### Using `break` to exit a loop
- `break` stops the loop immediately and continues after it.
- Example: stop once the square exceeds 150.

```python
for n in range(42):
    print(f"{n} * {n} = {n * n}")
    if n * n > 150:
        break
# prints up to 13 * 13 = 169, then stops
```
