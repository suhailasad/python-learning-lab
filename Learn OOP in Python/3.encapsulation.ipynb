## What is Encapsulation?

Encapsulation means hiding the internal details of how an object works and only exposing what is necessary. It helps protect data and makes your code easier to use and maintain.

### Example: Bank Account
You do not want anyone to set the account balance directly. Instead, you expose methods like `deposit` and `withdraw`.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # __balance is "private"

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        else:
            print("Insufficient funds or invalid amount.")

    def get_balance(self):
        return self.__balance

# Usage
account = BankAccount("Alice", 100)
account.deposit(50)          # Deposited 50. New balance: 150
account.withdraw(70)         # Withdrew 70. New balance: 80
print(account.get_balance()) # 80

# Direct access is not allowed:
# print(account.__balance)   # This will give an error!
```

### Purpose
- Protects data: You cannot accidentally set a negative balance or bypass rules.
- Controls access: Only methods can change the balance, and they validate operations.

### Real-Time Benefit
Encapsulation ensures users cannot cheat or break the system by setting their balance directly. If you need to change how balance is calculated or stored, you update the class without touching all its callers. It makes code safer, more reliable, and easier to maintain.

### Summary
Encapsulation hides details and protects important data. It lets you control how data is changed, which is crucial in real-world programs like banking, games, and more.


````python
# Private method and private attribute example
class Robot:
    def __init__(self):
        self.__battery = 100

    def move(self, distance):
        if self.__has_power():
            self.__battery -= distance
            print(f"Moved {distance} units. Battery left: {self.__battery}")
        else:
            print("Battery empty!")

    def __has_power(self):  # Private method
        return self.__battery > 0

robot = Robot()
robot.move(10)       # Works fine
# robot.__has_power() # Error! Can't access private method from outside
````
